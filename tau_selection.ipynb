```python
df = pd.read_csv("results/Ribo_tau_selection_08_17_01_33_37.csv", header=1)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def compute_time_to_convergence(df, eps=1e-3, target_col='iter'):
  """Trova la prima iterazione o tempo in cui gap <= eps per ogni combinazione."""
  records = []
  for (algo, tau), group in df.groupby(['algorithm', 'tau']):
    conv = group[group['gap'] <= eps]
    if not conv.empty:
      val = conv[target_col].iloc[0]
    else:
      val = np.nan  # o group[target_col].max() se non ha raggiunto la soglia
    records.append({'algorithm': algo, 'tau': tau, 'conv_val': val})
  return pd.DataFrame(records)


# Visualizzazione
conv_df = compute_time_to_convergence(df, eps=1e-3, target_col='iter')

plt.figure(figsize=(9, 5))
for algo, group in conv_df.groupby('algorithm'):
  plt.plot(
      group['tau'],
      group['conv_val'],
      marker='o',
      linewidth=2,
      label=algo,
  )

plt.xscale('log')
plt.yscale('log')
plt.xlabel(r'$\tau$ (L1 Radius)', fontsize=12)
plt.ylabel(r'Iterations to reach Gap $\le 10^{-3}$', fontsize=12)
plt.title(
    r'Convergence Speed across $\tau$ values ($\varepsilon = 10^{-3}$)',
    fontsize=14,
)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
